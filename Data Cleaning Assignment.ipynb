Q2. Handling Missing Age
Scenario:Age is critical for medical analysis, but some values are missing.
Task:
A)Replace missing Age values with an appropriate method
B)Justify your choice (mean/median)

```python
#A)
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)

print(df['Age'].isnull().sum())
```

B)
I used median because:
The dataset contains ages ranging from 0 to 99
Median is less affected by extreme values/outliers
It preserves the central tendency of patient ages better in medical datasets
Although the mean (49.60) is close to the median (50), median is generally preferred for healthcare data cleaning because it is more robust.

Q3. Handling Missing Treatment Cost
Scenario:Treatment cost is highly skewed due to expensive treatments.
Task:
A)Handle missing Treatment_Cost values
B)Choose the correct imputation method and explain why

```python
#A)
print(df['Treatment_Cost'].isnull().sum())
median_cost = df['Treatment_Cost'].median()
print("Median Treatment Cost:", median_cost)
df['Treatment_Cost'].fillna(median_cost, inplace=True)
print(df['Treatment_Cost'].isnull().sum())
```

B)
I used median instead of mean because:
The dataset is highly skewed
Some patients have very expensive treatments (outliers)
Mean gets heavily affected by extreme values
Median represents the typical treatment cost more accurately
Since skewness is very high, median imputation is the most reliable method for this dataset.

Q6. Outlier Detection (Treatment Cost)
Scenario:Extreme treatment costs are affecting analysis.
Task:
Detect outliers using IQR method
Display number of outliers

```python
Q1 = df['Treatment_Cost'].quantile(0.25)
Q3 = df['Treatment_Cost'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Treatment_Cost'] < lower_bound) |(df['Treatment_Cost'] > upper_bound)]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Number of Outliers:", outliers.shape[0])
```

Q7. Outlier Treatment
Scenario:The business team wants to retain all records.
Task:
Apply capping (Winsorization) on Treatment_Cost
Use 5th and 95th percentile

```python
lower_cap = df['Treatment_Cost'].quantile(0.05)
upper_cap = df['Treatment_Cost'].quantile(0.95)
print("5th Percentile:", lower_cap)
print("95th Percentile:", upper_cap)
df['Treatment_Cost'] = df['Treatment_Cost'].clip(
    lower=lower_cap,
    upper=upper_cap)
print(df['Treatment_Cost'].describe())
```

Q8. Transformation
Scenario:Treatment cost is highly skewed.
Task:
Apply log transformation
Create a new column
Compare before vs after distribution

```python
df['Log_Treatment_Cost'] = np.log1p(df['Treatment_Cost'])
before_skew = df['Treatment_Cost'].skew()
after_skew = df['Log_Treatment_Cost'].skew()
print("Skewness Before Log Transformation:", before_skew)
print("Skewness After Log Transformation:", after_skew)
```
