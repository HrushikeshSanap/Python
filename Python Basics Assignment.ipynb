Question 2 : Explain the difference between List and Tuple in Python.
Answer:-
In Python, both List and Tuple are used to store collections of items, but they differ in key ways.
A List is mutable/changeable, meaning its elements can be changed, added, or removed after creation. It is defined using square brackets [] and is commonly used when data needs modification.
A Tuple is immutable/unchangeable, so its elements cannot be changed once defined. It is created using parentheses () and is preferred when data should remain constant.

Question 3 : What is a function in Python? Why are functions useful? 
Answer:-
A function in Python is a block of reusable code that performs a specific task and runs when it is called. It is defined using the def keyword and can take inputs (parameters) and return outputs.
Functions are useful because they promote code reusability, avoiding repetition of the same logic. They improve readability and organization by breaking a program into smaller parts.

```python
#Question 5 : Write a Python program to check whether a number is even or odd using conditional statements. 
#Answer:-
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(number, "is even.")
else:
    print(number, "is odd.")
```

```python
#Question 8 : Write a Python program to remove duplicate values from a list using a set.
#Answer:-
my_list = [1, 2, 3, 4, 2, 5, 1, 6]
unique_values = set(my_list)
print(list(unique_values))
```

```python
#Question 9 : Write a function that returns the square of a number.     
#Answer:-
def square(num):
    return num ** 2
number = float(input("Enter a number to find its square: "))
result = square(number)
print(result)
```

```python
#Question 10 : Write a Python program to count how many times a number appears in a list. Example List [2,3,4,2,5,2].
#Answer:-
my_list = [2, 3, 4, 2, 5, 2]
number_to_count = int(input("Enter a number to count its occurrences: "))
count = my_list.count(number_to_count)
print(f"The number {number_to_count} appears {count} times in the list.")
```
