Q7. Build Preprocessing Pipeline
Create a pipeline that:
Applies encoding to categorical columns
Applies scaling to numerical columns
Use:
ColumnTransformer
Pipeline

```python
#Answer:-
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import pandas as pd
df=pd.read_csv("C:\\Users\\ABHISHEK\\OneDrive\\Desktop\\Hrushikesh\\PW SKILLS ASSIGNMENTS\\Python\\employee_productivity_dataset.csv")
categorical_cols = ['Gender', 'Department', 'Work_Mode', 'Location']
numerical_cols = [
    'Age',
    'Salary',
    'Hours_Worked_Per_Week',
    'Projects_Completed',
    'Performance_Score']
categorical_pipeline = Pipeline([('encoder', OneHotEncoder(handle_unknown='ignore'))])
numerical_pipeline = Pipeline([('scaler', StandardScaler())])
preprocessor = ColumnTransformer([('cat', categorical_pipeline, categorical_cols),('num', numerical_pipeline, numerical_cols)])
processed_data = preprocessor.fit_transform(df)
print("Processed Data Shape:", processed_data.shape)
print(processed_data)

```

Q8. Apply Pipeline
Apply the pipeline on the dataset.
Display:
Transformed dataset
Shape of final dataset

```python
#Answer:-
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import pandas as pd
df=pd.read_csv("C:\\Users\\ABHISHEK\\OneDrive\\Desktop\\Hrushikesh\\PW SKILLS ASSIGNMENTS\\Python\\employee_productivity_dataset.csv")
categorical_cols = ['Gender', 'Department', 'Work_Mode', 'Location']
numerical_cols = [
    'Age',
    'Salary',
    'Hours_Worked_Per_Week',
    'Projects_Completed',
    'Performance_Score'
]

categorical_pipeline = Pipeline([('encoder', OneHotEncoder(handle_unknown='ignore'))])
numerical_pipeline = Pipeline([('scaler', StandardScaler())])
preprocessor = ColumnTransformer([('cat', categorical_pipeline, categorical_cols),('num', numerical_pipeline, numerical_cols)])
transformed_data = preprocessor.fit_transform(df)
encoded_columns = preprocessor.named_transformers_['cat'] \
    .named_steps['encoder'] \
    .get_feature_names_out(categorical_cols)
final_columns = list(encoded_columns) + numerical_cols
transformed_df = pd.DataFrame(transformed_data,columns=final_columns)
print(transformed_df.head())
print("Shape of Final Dataset:", transformed_df.shape)
```

Q9. Conceptual Question:-Why is scaling important in Python?
Answer:-
Scaling is important in Python because it converts numerical data into a similar range. Features like Salary and Age may have very different values, and large values can dominate machine learning models. Scaling improves model accuracy, helps algorithms train faster, and gives better results in algorithms like KNN, SVM, Linear Regression, and K-Means. It also makes distance calculations more accurate.

Q10. Conceptual Question Why do we convert categorical data into numerical form?
Answer:-
Converting categorical data into numerical form is done because machine learning algorithms cannot understand text values directly. Algorithms work only with numbers and mathematical calculations. Converting categories like Gender, Department, or Work_Mode into numerical values helps models process the data correctly and improve prediction accuracy. Common methods used are Label Encoding and One-Hot Encoding.
