Create a Pandas DataFrame using a Python dictionary with the following columns:Product,Region,Sales,Quantity

Q9. Descriptive Statistics
Generate summary statistics using:
describe()
From the result identify:
1)Mean sales value
2)Maximum quantity sold
3)Minimum sales value

```python
summary = df.describe()
print(summary)
```

```python
#1)
print("Mean Sales Value:", summary.loc['mean', 'Sales'])
```

```python
#2)
print("Maximum Quantity Sold", summary.loc['max','Quantity'])
```

```python
#3)
print("Minimum sales value", summary.loc['min','Sales'])
```
