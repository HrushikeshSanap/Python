Q4 Regional Sales Performance
Calculate the total sales generated by each region.
Then identify:
Which region generated the highest revenue?

```python
regional_sales=(df.groupby('Region')['Sales'].sum())
print(regional_sales)

highest_revenue_region= regional_sales.idxmax()
highest_revenue_value= regional_sales.max()
print("Highest revenue region",highest_revenue_region )
print("Highest revenue value",highest_revenue_value )
```

Q5 Product Performance Analysis
Calculate the total quantity sold for each product.
Which product appears to be the most popular?

```python
Product_Quantity=(df.groupby('Product')['Quantity'].sum())
print(Product_Quantity)
Most_Popular_Product= Product_Quantity.idxmax()
Highest_Quantity= Product_Quantity.max()
print("Most Popular Product:",Most_Popular_Product)
print("Highest Quantity Sold:",Highest_Quantity)
```

Q7 Pivot Table Creation
Create a pivot table showing:
Rows → Region
Columns → Product
Values → Total Sales
Which region sells the most laptops?

```python
pivot_table= pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='sum')
print(pivot_table)
Most_Laptop_sales_region= pivot_table['Laptop'].idxmax()
Most_Laptop_sales_Value=pivot_table['Laptop'].max()
print("Most Laptop Region:",Most_Laptop_sales_region)
print("Sales_Value:",Most_Laptop_sales_Value)
```

Q8 Profitability Analysis
Calculate the average profit for each product category.
Which category is most profitable?

```python
average_profit= df.groupby('Category')['Profit'].mean()
print("Average Profit for Each Category",average_profit)
most_profitable_category = average_profit.idxmax()
highest_average_profit = average_profit.max()
print("Most Profitable Category:", most_profitable_category)
print("Highest Average Profit:", highest_average_profit)
```

Q9 Discount Impact Study
Analyze the relationship between discount and profit.
Tasks:
1)Calculate the average profit for each discount level
2)Identify whether higher discounts lead to lower profit

```python
discount_profit = df.groupby('Discount')['Profit'].mean()
print("Average Profit for Each Discount Level:")
print(discount_profit)
```

```python
#2)
if discount_profit.is_monotonic_decreasing:
    print("Higher discounts generally lead to lower profit.")
else:
    print("Higher discounts do not always lead to lower profit.")
```

2)Yes high discount lead to decrease in profit.
Moderate discounts may help increase sales, but excessive discounts can reduce overall profit.

Q10 End-to-End Data Wrangling Task
Perform the following workflow:
Filter the dataset where Sales > 3000
Group data by Region and Product
Calculate total sales
Reshape the result into a pivot table
Explain the business insights obtained from this analysis.

```python
#1)
filtered_data = df[df['Sales'] > 3000]
```

```python
#2)
grouped_sales = filtered_data.groupby(['Region', 'Product'])
print(grouped_sales)
```

```python
#3)
Total_sales=grouped_sales['Sales'].sum()
print(Total_sales)
```

```python
#4)
pivot_result = Total_sales.unstack()
print("Pivot Table:")
print(pivot_result)
```

5)
The pivot table clearly shows which products generate the highest sales in each region.
eg:-Tablet in East
Regions(eg:-East) with consistently high sales indicate strong market performance.
Products with high sales across multiple regions may require increased inventory and marketing focus.
Low-performing products or regions may need promotional strategies or pricing improvements.
e.g:-Printer in east;Tablet in North; Headphones in South and Laptop in west.
This analysis helps businesses make region-wise product planning and sales decisions.
